```python
from pathlib import Path

BASE_DIR = Path("/content/drive/MyDrive/KLTN_hand_gesture L2")
DATA_YAML = BASE_DIR / "gesture.yaml"
MODEL_PRETRAINED = BASE_DIR / "yolo12m.pt"
RUN_NAME = "gesture_v2_1000img"

print("YAML:", DATA_YAML)
print("Pretrained model:", MODEL_PRETRAINED)
print("Tồn tại YAML?", DATA_YAML.exists())
print("Tồn tại yolo12m.pt?", MODEL_PRETRAINED.exists())
```

```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/KLTN_hand_gesture L2/yolo12m.pt")

model.train(
    data="/content/drive/MyDrive/KLTN_hand_gesture L2/gesture.yaml",
    epochs=80,
    imgsz=640,
    batch=16,
    workers=2,
    project="/content/drive/MyDrive/KLTN_hand_gesture L2/runs",
    name="gesture_v2_1000img"
)
```
